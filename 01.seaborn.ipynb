Cómo se distribuye el largo del pico de los pinguinos?

```python
df.body_mass_g.median()
```

```python
df.body_mass_g.quantile(0.75)
```

```python
sns.histplot(x=df.body_mass_g, bins=15)
plt.axvline(x=df.body_mass_g.quantile(0.10), label="q10", color='green')
plt.axvline(x=df.body_mass_g.median(), label="median", color='orange')
plt.axvline(x=df.body_mass_g.quantile(0.75), label="q75", color='red')
plt.legend()
```

```python
for s in df.species.unique():
    df_temp = df[df.species == s]
    corr = df_temp[["body_mass_g", "bill_depth_mm"]].corr().iloc[0, 1]

    print(f"la correlación entre peso y pico para la especie {s} es {round(corr, 2)}")
```
