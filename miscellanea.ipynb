crea una función que
- recibe una palabra
- devuelve: la palabra en mayúscula si empieza por 'a' o 'A', y si no, la palabra en minúscula

```python
def imprimir_nombre(nombre):
    if (nombre[0] == 'a') or (nombre[0] == 'A'):
        return nombre.upper()
    else:
        return nombre.lower()
```

```python
imprimir_nombre("PEPITO")
```

```python
imprimir_nombre("juan")
```

```python
imprimir_nombre("alan")
```

```python
imprimir_nombre("Arturo")
```

```python
def tirar_hasta_seis():
    contador = 0
    
    while True:
        contador += 1
        dado = random.choice([1, 2, 3, 4, 5, 6])
        # print(dado)
    
        if dado == 6:
            break

    return contador
```

```python
tirar_hasta_seis()
```

```python
cuaderno = []

for day in range(365):
    tiradas_del_dia = tirar_hasta_seis()

    cuaderno.append(tiradas_del_dia)
```

```python
max(cuaderno)
```

```python
min(cuaderno)
```

```python
round(sum(cuaderno) / len(cuaderno), 1)
```

Un número es _guay_ si es divisible por 2 o por 7. Calcula la suma de todos los números _guays_ entre 0 y 100.

```python
15 % 7 == 0
```

```python
str(89)
```

```python
def is_guay(n):
    if (n % 7 == 0) or (n % 2) == 0:
        return True
    else:
        return False
```

```python
is_guay(8)
```

```python
is_guay(11)
```

```python
is_guay(77)
```

```python
total = 0

for n in range(101):
    if is_guay(n):
        total += n
```

```python
total
```

Un número es _siético_ si contiene un 7. Calcula la suma de todos los números _siéticos_ entre 0 y 1000.

```python
'a' in 'Manuel'
```

```python
'7' in str(72)
```

```python
'7' in str(88)
```

```python
def is_sietico(n):
    if '7' in str(n):
        return True
    else:
        return False
```

```python
is_sietico(17)
```

```python
is_sietico(22)
```

```python
total = 0

for n in range(1001):
    if is_sietico(n):
        total += n
```
